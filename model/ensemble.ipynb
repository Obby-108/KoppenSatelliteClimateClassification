```python
import h5py
import numpy as np
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import classification_report
from sklearn.preprocessing import StandardScaler
```

```python
SEED = 42
# Import data via h5py compressed file
with h5py.File('svm_ready_data.h5', 'r') as hf:
    # Extract into standard numpy arrays
    X_train = np.array(hf['X_train'])
    y_train = np.array(hf['y_train'])
    X_test = np.array(hf['X_test'])
    y_test = np.array(hf['y_test'])
print("Successfully retrieved 5-statistic/12-class/60-feature data.")

# Standardize the data
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
# Define RF model
rf_model = RandomForestClassifier(
    n_estimators=500,
    max_depth=20,
    n_jobs=-1,  # Use all CPU cores
    random_state=SEED,
    class_weight='balanced'
)

# Fit model to training data
rf_model.fit(X_train, y_train)

# Evaluate on test data
outputs = rf_model.predict(X_test)
print(classification_report(y_test, outputs))
```

```python
# Define GBC model
gbc_model = GradientBoostingClassifier(
    n_estimators=500,
    learning_rate=0.1,
    max_depth=5,
    subsample=0.8,
    validation_fraction=0.1,
    n_iter_no_change=5,
    verbose=1,
    random_state=SEED
)

# Fit model on training data
gbc_model.fit(X_train, y_train)

# Evaluate on test data
outputs = gbc_model.predict(X_test)
print(classification_report(y_test, outputs))
```
